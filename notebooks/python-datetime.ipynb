- The Problem: To a computer, the text "2023-01-01" is just a string of characters, like "Hello World". It doesn't know that "2023-01-02" comes one day after it.
- The Solution: We convert these strings into Datetime Objects. This allows us to do math on dates (e.g., Date_B - Date_A = 3 days) and handle business days vs. weekends.

```python
import datetime
import pandas as pd
import numpy as np
```

```python
# Find the difference in days from 2 dates
# using datetime module
datefirst = 2026, 1, 1
datesecond = 2026, 12, 31

# * means unpacking the tuple
date1 = datetime.datetime(*datefirst)
date2 = datetime.datetime(*datesecond)
date_diff = (date2 - date1).days
print(f"Date difference: {date_diff} days")
# print date_diff in months and days
months = date_diff // 30
days = date_diff % 30
print(f"Date difference: {months} months and {days} days")
# exclude Saturdays and Sundays in months and days calculation
weekdays = np.busday_count(date1.date(), date2.date())
months_weekdays = weekdays // 30
days_weekdays = weekdays % 30
print(f"Date difference (weekdays only): {months_weekdays} months and {days_weekdays} days")
```

```python
# using pandas
date1 = pd.to_datetime('2025-12-08')
date2 = pd.to_datetime('2026-04-30')
date_diff = (date2 - date1).days
print(f"Date difference: {date_diff} days")
```

```python
# using numpy
date1 = np.datetime64('2025-12-08')
date2 = np.datetime64('2026-04-30')
date_diff = (date2 - date1).astype(int)
print(f"Date difference: {date_diff} days")
```
